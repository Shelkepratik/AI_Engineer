Write a calculator program with four separate functions: add(), sub(), mul() and div(). Each function should take two numbers and return the result. Call all four functions with the numbers 20 and 5.

```python
def add(a,b):
    return a+b

def sub(a,b):
    return a-b

def mul(a,b):
    return a*b

def div(a,b):
    return a/b

result1=add(20,5)

result2=sub(20,5)

result3=mul(20,5)

result4=div(20,5)


print(result1)

print(result2)

print(result3)

print(result4)
```
