```python
num1=int(input("Enter the first number: "))
num2=int(input("Enter the second number: "))
```

```python
print("num1: ",num1)
print("num2: ",num2)
try:
    result=num1/num2
    print(result)
except ZeroDivisionError as err:
    print(err)
```

2)overflow error:-overflow error raised when a floating point arithmetic result is too large to be represented

```python
import math
math.exp(710)
```

```python
import math 
try:
    value=math.exp(710)
    print(value)
except OverflowError as err:
    print("overflow exception received: ",err)
```
